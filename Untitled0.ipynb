```python
import torch
from torch import nn
import matplotlib.pyplot as plt

# 1. PyTorch versiyasi
print(torch.__version__)
```

```python
# 2. CPU yoki GPU tanlash
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)
```

```python
# 3. Weight va Bias
weight = 0.7
bias = 0.3

# 4. Data yaratish
start = 0
end = 1
step = 0.02

X = torch.arange(start, end, step).unsqueeze(dim=1)
y = weight * X + bias

# Liner Regression: y = wx + b
print("X:", X)
print("\n y:", y)

# 5. Train/Test Split
train_split = int(0.8 * len(X))

X_train, y_train = X[:train_split], y[:train_split]
X_test, y_test = X[train_split:], y[train_split:]

print(len(X_train), len(y_train), len(X_test), len(y_test))
```

```python
def plot_predictions(train_data=X_train,
                     train_labels=y_train,
                     test_data=X_test,
                     test_labels=y_test,
                     predictions=None):

    plt.figure(figsize=(10, 7))

    # Training data — Blue
    plt.scatter(train_data, train_labels,
                c='b', s=4, label='Training data')

    # Testing data — Green
    plt.scatter(test_data, test_labels,
                c='g', s=4, label='Testing data')

    # Predictions — Red
    if predictions is not None:
        plt.scatter(test_data, predictions,
                    c='r', s=4, label='Predictions')

    plt.legend(prop={'size': 14})

plot_predictions()
```

```python
import torch
from torch import nn

# 1. Linear Regression Model yaratish
class LinearRegressionModelV2(nn.Module):

    def __init__(self):
        super().__init__()

        self.linear_layer = nn.Linear(
            in_features=1,
            out_features=1
        )

    # 2. Forward Pass
    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.linear_layer(x)

# 3. Model yaratish
torch.manual_seed(42)

model_1 = LinearRegressionModelV2()

# 4. Model parametrlarini ko'rish
model_1, model_1.state_dict()
```

```python
# 1. Loss Function — Xatolikni hisoblash
loss_fn = nn.L1Loss()  # MAE: Mean Absolute Error

# 2. Optimizer — Model parametrlarini yaxshilash
optimizer = torch.optim.SGD(
    params=model_1.parameters(),
    lr=0.01
)
```

```python
# 3. Training'dan oldingi Predictions
model_1.eval()

with torch.inference_mode():
    y_preds_early = model_1(X_test)

# 4. Predictions Visualization
plot_predictions(predictions=y_preds_early)
plt.title("Predictions before training")
```

```python
# 1. Epochs — O'qitish takrorlanishlari
epochs = 1000

# 2. Data'ni CPU yoki GPU'ga o'tkazish
model_1.to(device)
X_train = X_train.to(device)
X_test = X_test.to(device)
y_train = y_train.to(device)
y_test = y_test.to(device)

for epoch in range(epochs):

    # TRAINING
    model_1.train()

    # 1. Forward Pass — Bashorat
    y_pred = model_1(X_train)

    # 2. Calculate Loss — Xatolik
    loss = loss_fn(y_pred, y_train)

    # 3. Zero Grad — Eski gradientlarni tozalash
    optimizer.zero_grad()

    # 4. Backward — Gradientlarni hisoblash
    loss.backward()

    # 5. Optimizer Step — Parametrlarni yangilash
    optimizer.step()

    # TESTING
    model_1.eval()

    with torch.inference_mode():
        test_pred = model_1(X_test)
        test_loss = loss_fn(test_pred, y_test)

    # Har 100 epoch'da natijani chiqarish
    if epoch % 100 == 0:
        print(f"Epoch: {epoch} | Train loss: {loss} | Test loss: {test_loss}")
```

```python
# 1. Modelni Evaluation rejimiga o'tkazish
model_1.eval()

# 2. Test Data orqali bashorat qilish
with torch.inference_mode():
    y_preds = model_1(X_test)

# 3. Natijani ko'rish
y_preds
```

```python
# Predictions'ni CPU'ga o'tkazish va grafikda ko'rsatish
plot_predictions(predictions=y_preds.cpu())
```
